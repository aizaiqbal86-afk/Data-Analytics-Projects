```python
import numpy as np
import pandas as pd
x = [10, 20, 30, 40]
y = [17, 19, 21, 23]   # added one value
data = {'x': x, 'y': y}
df = pd.DataFrame(data)
corr = np.corrcoef(x, y)[0,1]
print(df)
print("Correlation:", corr)
```

```python
data = {'x':[10,20,30,40], 'y':[15,25,35,45]}
df = pd.DataFrame(data)
corr = df['x'].corr(df['y'])
print(corr)
```
