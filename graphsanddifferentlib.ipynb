# By using different libraries finding mean,median,std,mode and making histogram of random data:

```python
import numpy as np
from scipy import stats
import pandas as pd
import matplotlib.pyplot as plt
k=np.random.uniform(0.0,5.0,50)
print(k)
df=pd.DataFrame(k,columns=["randomValues"])
print("mean:",np.mean(k))
print("median:",np.median(k))
print("mode:",stats.mode(k))
print("sum:",sum(k))
print("describe data:",df.describe())
plt.hist(k,10)
plt.show()
plt.hist(k,20)
plt.show()
```

# Making histogram using matplotlib library:

```python
import matplotlib.pyplot as plt
import pandas as pd
data = [1,2,2,3,3,3,4,4,5,8,7,9]
plt.hist(data, bins=8) 
plt.show()
df = pd.DataFrame(data,columns=["Values"]) #convert data into table
print("INFO:")
print(df.info())
print("\nDESCRIBE:") 
print(df.describe())
```
