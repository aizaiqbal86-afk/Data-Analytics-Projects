# Data validation

```python
import pandas as pd
import numpy as np
file_path="employee.xlsx"
df=pd.read_excel(file_path)
print(df.head())
```

# Ensure column has no missing values

```python
assert df['Employee ID'].notnull().all(), "Employee ID column contains missing values"
```

# Validate numeric range

```python
assert df['Total Working Days'].between(0, 100).all(), "Total Working Days must be between 0 and 120"
```

# Validate allowed categories

```python
valid = {'HR', 'IT', 'Finance','Admin','Marketing','Accounts'}
assert set(df['Department']).issubset(valid), "Invalid Department category found"
```

# Validate unique values

```python
assert df['Employee ID'].is_unique, "Duplicate Employee ID  found"
```

# Validate date column format

```python
df['date'] = pd.to_datetime(df['date'], errors='raise')
```

# Ensure numeric columns contain only numbers

```python
assert pd.api.types.is_numeric_dtype(df['Total Working Days']), "Total Working Days contains non-numeric data"
```
