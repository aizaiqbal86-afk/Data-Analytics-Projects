# XML 

```python
import pandas as pd

# Read XML file
df = pd.read_xml("data (1).xml")

# Show data
print(df.head())
```

```python
import xml.etree.ElementTree as ET

# Load XML file
tree = ET.parse("data (1).xml")
root = tree.getroot()

# Print root tag
print(root.tag)

# Loop through elements
for child in root:
    print(child.tag, child.attrib)
```
