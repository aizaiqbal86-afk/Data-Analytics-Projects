# Machine Learning Algorithms on Iris_data_Set:

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, mean_squared_error

# 1. Load the dataset
Data = pd.read_csv("iris_dataset (1).csv")

# 2. Encode the target column
en = LabelEncoder()
y = en.fit_transform(Data["target"])  # make sure 'target' is the column name

# 3. Feature matrix
x = Data.drop("target", axis=1)

# 4. Normalize/scale features
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)

# 5. Split dataset into training and test sets
x_train, x_test, y_train, y_test = train_test_split(
    x_scaled, y, train_size=0.8, test_size=0.2, random_state=123
)

# 6. Train models
# Logistic Regression
log_model = LogisticRegression(max_iter=200)
log_model.fit(x_train, y_train)

# Support Vector Machine (SVM)
svm_model = SVC(kernel="linear", C=1)
svm_model.fit(x_train, y_train)

# Random Forest
rf_model = RandomForestClassifier()
rf_model.fit(x_train, y_train)

# 7. Predict using Random Forest (you can switch to svm_model or log_model)
y_pred = rf_model.predict(x_test)

# 8. Evaluation
print("\nPredicted labels for the test set:\n", y_pred)
print("Actual labels for test: ", y_test)

accuracy = accuracy_score(y_test, y_pred)
print("\nAccuracy of the model: {:.2f}%".format(accuracy*100))

print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))

mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
print(f"\nRoot Mean Squared Error (RMSE): {rmse:.2f}")
```

# Wine Data set

```python
# ======================================================
# WINE DATASET COMPLETE MACHINE LEARNING ASSIGNMENT
# ======================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix, mean_squared_error

# ------------------------------------------------------
# 1. Load Dataset
# ------------------------------------------------------
Data = pd.read_csv("wine.csv")

print("Dataset Preview:\n", Data.head())

# ------------------------------------------------------
# 2. Data Type Conversion (if needed)
# ------------------------------------------------------
# convert all columns to numeric safely
Data = Data.apply(pd.to_numeric, errors='coerce')

# fill missing values if created during conversion
Data = Data.fillna(Data.mean())

print("\nData Types:\n", Data.dtypes)

# ------------------------------------------------------
# 3. EDA (Exploratory Data Analysis)
# ------------------------------------------------------

# Histogram
Data.hist(figsize=(12,10))
plt.suptitle("Feature Distributions")
plt.show()

# Heatmap
plt.figure(figsize=(10,8))
sns.heatmap(Data.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()

# Scatter Plot
sns.scatterplot(x="Alcohol", y="Color.int", data=Data)
plt.title("Alcohol vs Color Intensity")
plt.show()

# ------------------------------------------------------
# 4. Feature & Target Selection
# ------------------------------------------------------
# Regression target
y = Data['Proline']

# Features
x = Data.drop('Proline', axis=1)

# ------------------------------------------------------
# 5. NORMALIZATION (VERY IMPORTANT)
# ------------------------------------------------------
scaler = StandardScaler()

x_scaled = scaler.fit_transform(x)

# ------------------------------------------------------
# 6. Train Test Split (Regression)
# ------------------------------------------------------
x_train, x_test, y_train, y_test = train_test_split(
    x_scaled, y,
    test_size=0.2,
    random_state=42
)

# ------------------------------------------------------
# 7. PCA (Dimensionality Reduction)
# ------------------------------------------------------
pca = PCA(n_components=5)

x_train = pca.fit_transform(x_train)
x_test = pca.transform(x_test)

print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)

# ------------------------------------------------------
# 8. Classification Target Creation
# ------------------------------------------------------
# Convert continuous target into categories
y_class = pd.qcut(y, q=3, labels=[0,1,2])

x_train_c, x_test_c, y_train_c, y_test_c = train_test_split(
    x_scaled, y_class,
    test_size=0.2,
    random_state=42
)

# ------------------------------------------------------
# 9. Logistic Regression (Classification)
# ------------------------------------------------------
log_model = LogisticRegression(max_iter=5000)

log_model.fit(x_train_c, y_train_c)

y_pred_log = log_model.predict(x_test_c)

print("\nLogistic Regression Accuracy:",
      accuracy_score(y_test_c, y_pred_log)*100)

print("Confusion Matrix (Logistic):")
print(confusion_matrix(y_test_c, y_pred_log))

# ------------------------------------------------------
# 10. SVM Classification
# ------------------------------------------------------
svm = SVC(kernel='linear')

svm.fit(x_train_c, y_train_c)

y_pred_svm = svm.predict(x_test_c)

print("\nSVM Accuracy:",
      accuracy_score(y_test_c, y_pred_svm)*100)

print("Confusion Matrix (SVM):")
print(confusion_matrix(y_test_c, y_pred_svm))

# ------------------------------------------------------
# 11. Linear Regression
# ------------------------------------------------------
reg = LinearRegression()

reg.fit(x_train, y_train)

y_pred_reg = reg.predict(x_test)

print("\nPredicted Values:", y_pred_reg[:5])
print("Actual Values:", y_test[:5])

# ------------------------------------------------------
# 12. MSE & RMSE
# ------------------------------------------------------
mse = mean_squared_error(y_test, y_pred_reg)
rmse = np.sqrt(mse)

print("\nMSE:", mse)
print("RMSE:", rmse)

print("\nTotal PCA Components Used:", pca.n_components_)
```
